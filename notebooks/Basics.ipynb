```kotlin
data class Person(val name: String, val age: Int)

val people = listOf(Person("Ana", 31), Person("Ben", 25), Person("Chloe", 40))
people.sortedBy { it.age }
```

```kotlin
people.joinToString(",")
```

```kotlin
people.joinToString(",") {it.name}
```

```kotlin
people.joinToString(",", prefix="[", postfix="]") {it.name}
```

```kotlin
people.map { it.name }.joinToString("-")    // same idea, split in two steps
```
